```python
df['agent']=df['agent'].astype(int)
```

```python
df['arrival_date'] = pd.to_datetime(
    df['arrival_date_year'].astype(str) + '-' +
    df['arrival_date_month'] + '-' +
    df['arrival_date_day_of_month'].astype(str)
)
```

```python
df
```

```python
df['total_stay']=df['stays_in_weekend_nights']+df['stays_in_week_nights']
```

```python
df['total_guest']=df['adults']+df['children']+df['babies']
```

```python
df['revenue']=df['revenue'].round(2)
```

```python
df['revenue']=df['adr']*df['total_stay']
```

```python
df['revenue']=df['revenue'].round(2) #rounded with 2 decimal
```

```python
df['revenue'].sum()
```

```python
df['revenue'] = df.apply(            #applied a condition for each row with lambda function->if booking canceled=1->revenue=0,not canceled=revenue
    lambda row: row['adr'] * row['total_stay']
    if row['is_canceled'] == 0 else 0,
    axis=1
)
```
